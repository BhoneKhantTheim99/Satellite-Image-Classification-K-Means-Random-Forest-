```python
import geopandas as gpd
# Read and Load the KML
gdf = gpd.read_file("../region_of_interest/rare-earth sites Kachin.kml")

# Dropping Z coordinates (GEE doesn't support Z)
gdf['geometry'] = gdf['geometry'].apply(
    lambda geom: geom.__class__(
        [(x, y) for x, y, *_ in geom.exterior.coords]
    ) if geom.geom_type == 'Polygon' else geom
)

# Making sure CRS is WGS84
gdf = gdf.to_crs(epsg=4326)

gdf.to_file("roi_polygon.geojson", driver="GeoJSON")   # Export to GeoJSON
gdf.to_file("roi_polygon.shp")                         # Export to shape file
print(gdf)
```
